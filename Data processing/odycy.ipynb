```python
!pip install --quiet https://huggingface.co/chcaa/grc_odycy_joint_trf/resolve/main/grc_odycy_joint_trf-0.7.0-py3-none-any.whl

# Using spacy.load().
#import spacy
#nlp = spacy.load("grc_odycy_joint_trf")

# Importing as module.
import grc_odycy_joint_trf
nlp = grc_odycy_joint_trf.load()
```

```python
def pos_tagger(text):
  doc = nlp(text)
  results = []

  for token in doc:
    results.append(token.pos_)

  return results
```

```python
import pandas as pd
df = pd.read_csv('/content/drive/MyDrive/Tesi IUSS/brill_dictionary_processed.csv')

for i, row in df.iterrows():
  pos = row['part_of_speech']

  if pd.isna(pos) or str(pos).strip() == '':
      tags = pos_tagger(row['lemma_raw'])
      df.at[i, 'part_of_speech'] = tags[0]
      print(i)

df.to_csv('/content/drive/MyDrive/odycy.csv', index=False)
```
